```python
import chromadb

client = chromadb.Client()
#Client is the object throught which python communicates with the db
```

```python
collection = client.create_collection(
    name = "documents"
)
#Collection is where chroma will store the records 
#Think of this as a table in sql
```

```python
collection.add(
    ids=["1","2"],
    documents=[
        "Python is a programming language.",
        "Java is commonly used "
    ]
)
```

```python
collection.query(query_texts = ["C++ is a fun language"],
                 n_results= 2)
#You can check the similarity of one query string to another queyr string 
```
